```python
import numpy as np

rng = np.random.default_rng(3)
rain = np.round(rng.gamma(0.6, 9, 30), 1)  # 30 days of synthetic rainfall (mm)
print(f"Total: {rain.sum():.1f} mm, wet days: {(rain >= 2.5).sum()}")
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(6, 2.6))
ax.bar(range(1, 31), rain, color='#1d9e75')
ax.set_xlabel('Day')
ax.set_ylabel('Rainfall (mm)')
plt.show()
```
